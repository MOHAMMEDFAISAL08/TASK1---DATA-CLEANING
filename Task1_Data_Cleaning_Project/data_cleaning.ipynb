# Task 1: Data Cleaning & Structural Validation

```
import pandas as pd
df=pd.read_csv('data/raw_employee_data.csv')
df.head()
```

```
df=df.drop_duplicates()
df['Name']=df['Name'].fillna('Unknown').astype(str).str.strip().str.title()
df['Department']=df['Department'].fillna(df['Department'].mode()[0]).astype(str).str.title()
df['Salary']=df['Salary'].fillna(df['Salary'].median())
df['City']=df['City'].fillna('Unknown')
df
```

```
print(df.isnull().sum())
print(df.duplicated().sum())
```
